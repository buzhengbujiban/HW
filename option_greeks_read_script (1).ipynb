# Loading the minutely data for any one given day for all stocks

### initialize daily matrices for different variables

```python
date=20240531

out={}
for v in var_list:
    out[v]=np.full((data['numstocks'],len(hhmm_str)),np.nan,dtype=np.float32)
```

### read minute files for every minute and assign values to the matrices

```python
ticker2si=pd.DataFrame({'ticker':tickerNameMaps[date].keys(),'si':tickerNameMaps[date].values()})
for mi,hhmm in enumerate(hhmm_str):
    # standardized option info saved in three different files with 0,1,2 as postfix
    for i in [0,1,2]:
        results=pd.read_csv('/home/tqdata/data/nanex_option_derived/'+str(date)+'/nanex_stdop_greeks/option_minute.'+hhmm+'_'+str(i),sep='|',names=names,skiprows=1)
        results=results.merge(ticker2si,left_on='SYMBOL',right_on='ticker',how='inner')
        for v in var_list:
            out[v][results['si'].values,mi]=results[v].values
```

```python

def AR1generator(size):
    e_t = np.random.normal(size=size)
    f_t = np.zeros(size)
    for t in range(1, size):
        f_t[t] = 0.5 * f_t[t-1] + e_t[t]
    return f_t
N=30
n_samples=300
f_t = AR1generator(size=n_samples)    ### depending on your case
alphas = np.random.uniform(low=0.1, high=1.0, size=N)
u_it = np.random.normal(size=(n_samples, N))   ### depending on your case
correlated_errors = f_t[:, np.newaxis] * alphas[np.newaxis, :] + u_it
series_list = []
```

```python
for i in range(10):
    arma_model = sm.tsa.ArmaProcess(ar=np.r_[1, -0.3], ma=np.r_[1, 0.2])
    series = arma_model.generate_sample(nsample=n_samples, scale=1,
                                        distrvs=lambda size: correlated_errors[:size[0], i])
    series_list.append(series)
for i in range(10,20):
    arma_model = sm.tsa.ArmaProcess(ar=np.r_[1, -0.6], ma=np.r_[1, 0.3])
    series = arma_model.generate_sample(nsample=n_samples, scale=1,
                                        distrvs=lambda size: correlated_errors[:size[0], i])
    series_list.append(series)
for i in range(20,30):
    ar_params = np.random.uniform(0, 0.5, 1).tolist() + np.random.uniform(0, 0.2, 1).tolist()
    ma_params = np.random.uniform(0, 0.3, 1).tolist()
    ar = np.r_[1, -np.array(ar_params)]
    ma = np.r_[1, np.array(ma_params)]
    # 创建 ARMA(2,1) 模型
    arma_model = sm.tsa.ArmaProcess(ar, ma)
    series = arma_model.generate_sample(nsample=n_samples, scale=1,
                                        distrvs=lambda size: correlated_errors[:size[0], i])
    series_list.append(series)
```
